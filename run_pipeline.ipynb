# Adriatic One Health AMR indices (PORI and AMI): reproduction
Runs the full pipeline on `data/adriaticlast.xlsx` (about 1 minute), writes tables and figures to `results/` and checks key numbers against the manuscript.

```python
import os, subprocess, sys
env = dict(os.environ, AMR_XLSX='data/adriaticlast.xlsx', AMR_OUT='results')
r = subprocess.run([sys.executable, 'adriatic_amr_pipeline.py'], env=env, capture_output=True, text=True)
print(r.stdout[-3000:]); print(r.stderr[-2000:])
```

```python
r = subprocess.run([sys.executable, 'verifica.py'], capture_output=True, text=True)
print(r.stdout); print(r.stderr)
```
